```python
age = int(input("What is your age?"))
has_ticket = input("Do you have tickets yes/no?").strip(). lower()
print(has_ticket)

if age >= 18 and has_ticket == "yes":
  print("You may enter movie.")
elif age >= 18 and has_ticket == "no":
  print("Please purchase a ticket.")
elif age < 18 or has_ticket != "yes":
  print("Please purchase ticket or have parent purchase ticket for you.")
else:
  print("Please speak to staff.")
```

```python
raw = input("Enter an int between 1-10"). strip()
try:
  n = int(raw)

  if n < 1 or n > 10:
    raise ValueError("We require a value that is between 1 and 10")

except ValueError as e:
  print(f" Error {e} Your value was {raw}")
else:
  print("I am only printing after try not except  ")
```
